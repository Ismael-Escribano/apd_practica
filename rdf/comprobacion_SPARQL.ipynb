### Preparación del grafo a partir del archivo ttl generado

```python
from rdflib import Graph
import pandas as pd
g = Graph().parse("turismo.ttl")
```

### Comprobaciones:

#### Número de clases

```python
print('##### Número de clases:')

# Query the data in g using SPARQL
q = """
    SELECT (count(distinct ?class) as ?classes)
    WHERE {
        ?s a ?class .
    }
"""


for r in g.query(q):
    print(r["classes"])
```

#### Clases utilizadas

```python
print('##### Clases utilizadas:')

q = """
    SELECT distinct ?class
    WHERE {
        ?s a ?class .
    }
"""

for r in g.query(q):
    print(r["class"])
```

#### Número de propiedades

```python
print('##### Número de propiedades:')

# Query the data in g using SPARQL
q = """
    SELECT (count(distinct ?prop) as ?properties)
    WHERE {
        ?s ?prop ?o .
    }
"""

# Apply the query to the graph and iterate through results
for r in g.query(q):
    print(r["properties"])
```

#### Propiedades utilizadas

```python
print('##### Propiedades utilizadas:')


q = """
    SELECT distinct ?prop
    WHERE {
        ?s ?prop ?o .
    }
"""

for r in g.query(q):
    print(r["prop"])
```

#### Número de relaciones owl:sameAs

```python
print('##### Número de relaciones owl:sameAs:')
    
# Query the data in g using SPARQL
q = """
    SELECT (count(distinct ?sameAs) as ?total)
    WHERE {
        ?p owl:sameAs ?sameAs .
    }
"""

# Apply the query to the graph and iterate through results
for r in g.query(q):
    print(r["total"])
```

#### Relaciones owl:sameAs utilizadas

```python
print('##### Relaciones owl:sameAs utilizadas:')
    
# Query the data in g using SPARQL
q = """
    SELECT distinct ?sameAs
    WHERE {
        ?p owl:sameAs ?sameAs .
    }
"""

# Apply the query to the graph and iterate through results
for r in g.query(q):
    print(r["sameAs"])
```

#### Número de tripletas

```python
print('##### Número de tripletas:')
    
# Query the data in g using SPARQL
q = """
    SELECT (COUNT(*) as ?triples) 
    WHERE { ?s ?p ?o } 
"""

# Apply the query to the graph and iterate through results
for r in g.query(q):
    print(r["triples"])
```

#### Número de áreas administrativas

```python
print('##### Número de áreas administrativas:')
    
# Query the data in g using SPARQL
q = """
    SELECT (COUNT(distinct ?s) as ?AdministrativeArea) 
    WHERE {?s a schema:AdministrativeArea} 
"""

# Apply the query to the graph and iterate through results
for r in g.query(q):
    print(r["AdministrativeArea"])
```

#### Áreas administrativas utilizadas

```python
print('##### Áreas administrativas utilizadas:')
    
# Query the data in g using SPARQL
q = """
    SELECT distinct ?AdministrativeArea 
    WHERE {?AdministrativeArea a schema:AdministrativeArea} 
"""

# Apply the query to the graph and iterate through results
for r in g.query(q):
    print(r["AdministrativeArea"])
```

```python
query = """
PREFIX schema: <https://schema.org/>

SELECT ?comunidad ?pernoctaciones
WHERE {
    ?observacion a schema:Observation ;
        schema:variableMeasured "pernoctaciones" ;
        schema:observationAbout ?ccaa ;
        schema:value ?pernoctaciones .

    ?ccaa schema:name ?comunidad .
}
ORDER BY DESC(?pernoctaciones)
"""

resultados = g.query(query)

tabla = pd.DataFrame(
    [(str(r.comunidad), int(r.pernoctaciones))
     for r in resultados],
    columns=["Comunidad", "Pernoctaciones"]
)

tabla
```
